# For loops

```for``` loops are for iterating over a collection (list a list or tuple) or an iterator. The standard syntax for a ```for``` loop is:


````python
for value in collection:
    # do something with value 
````

You can advance a ```for``` loop to the next iteration, skipping the remainder of the block, using the ```continue``` keyword. Consider this code, which sums up intergers in a list and skips ```None``` values:


````python
sequence = [1, 2, None, 4, None, 5]
total = 0
for value in sequence:
    if value is None:
        continue
    total += value
````

A ```for``` loop can be exited altogether with a ```break``` keyword. This code sums elements of the list until a 5 is reached:

````python
sequence = [1, 2, 0, 4, 6, 5, 2, 1]
total_until_5 = 0

for value in sequence:
    if value == 5:
        break
    total_until_5 += value
````
