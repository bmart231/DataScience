SciPy (https://scipy.org) is a collection of packages addressing a number of foundational problems in scientific computing. Here are some of the tools it contains in its various modules. 

```python
import scipy

# Numerical integration routines and differential equation solvers
scipy.integrate 

# Linear algreba routines and matrix decompisitions extending beyond those provided in numpy.linalg
scipy.linalg

# Function optimizes (minimizers) and root finding algorithms 
scipy.optimize

# Signal processing tools
scipy.signal 

# Sparse matrices and sparse linear system solvers
scipy.sparse

# Wrapper around SPECFUN, a FORTAN library implementing many common mathematical functions, such
# as the gamma function
scipy.special

# Standard continous and descrete probability distributions (density function, sampelrs,
# continous distribution functions), various statistical tests, and more descriptive stats
scipy.stats 
```

## sickit-learn

Since the project's inception in 2007, sickit learn has become the premier general-purpose machine learning toolkit for python programmers. It includes submodels for models such as:

- Classification: SVM, nearest neighbors, random forest, logistic regresison, etc 
- Lasso, Ridge regression, etc

- Dimensionality reduction: PCA, feature selection, matrix factorization, etc
- Model Selection: Grid Search, cross-validation, metrics

## stats models

statsmodels is a statistical analysis package. Contains algorithms for classical (primarily frequentist) statistics and economics. This includes such submodules as:

- Regression Models:
- Analysis of Variance (ANOVA)
- Time Series analysis: AR, ARMA, ARIMA, VAR, and other models 
- Non Parametric moethods: Kernel density estimation, kernel regression
- Visualization of statistical model results

statsmodel is more focused on statistical inference, providing uncertainty estimates and $p$-values for parameters. sickit-learn, by contrast, is more prediction focused.
