# Pseudorandom Number Generator 

Then``numpy.random`` module supplements the built-in Python ``random`` module with functions for efficiently generating whole arrays of sample values from many kinds of probability distributions. For example, you can get a 4 x 4 array of samples from the standard normal distributions using ``numpy.random.standard_normal``:


```python
import numpy as np 

samples = np.random.standard_normal(size=(4,4))

samples
```

Python's built in ``random`` module, by contrast, samples only one value at a time. As you can see from this benchmark, ``numpy.ramdom`` is well over an order of magnitude faster for generating very large samples:

These random numbers are not truly random (rather, pseudorandom) but instead are generated by a configurable random number generator that determines deterministically what values are created. Functions like ``numpy.random.standard_normal`` use the ``numpy.random`` module's default random generator, but your code can be configured to use an explicit generator:

```python
rng = np.random.default_rng(seed = 12345)

data = rng.standard_normal((2, 3))
```

The ``seed`` argument is what determines the initial state of the generator, and the state changes each time the ``rng`` object is used to generate data. The generator object ``rng`` is also isolated from other code which might use the ``numpy.random`` module"
