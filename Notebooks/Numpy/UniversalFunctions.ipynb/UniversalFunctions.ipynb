# Universal Functions: Fast Element-Wise Array Functions

A universal function, or *ufunc*, is a function that performs element-wise operations on data in ndarrays. You can think of them as fast vectorized wrappers for simple functions that take one or more scalar values and produce one or more scalar results. 

Many ufuncs are simple element-wise transformations, like ``numpy.sqrt`` or ``numpy.exp``:

```python
import numpy as np 

arr = np.arange(10)

arr
```

```python
np.sqrt(arr)
```

```python
np.exp(arr)
```

These are reffered to as *unary* ufuncs. Others, such as ``numpy.add`` or ``numpy.maxium``, take two arrays (thus, *binary* ufuncs) and return a single array as the result:

```python
import numpy as np

x = np.random.standard_normal(8)

y = np.random.standard_normal(8)
```

```python
x, y
```

```python
np.maximum(x, y)
```

While not common, a ufunc can return multiple arrays. ``numpy.modf`` is one example: 

a vectorized version of the built-in Python ``math.modf``, it returns the fractional and integer parts of a floating point array:

```python
arr = np.random.standard_normal(7)*5

arr
```

```python
remainder, whole_part = np.modf(arr)

remainder
```

```python
whole_part
```
